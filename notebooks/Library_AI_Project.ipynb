```python
from scipy.sparse import csr_matrix

# Get unique users and books from the training data
train_users = train_interactions["Membership Number"].unique()
train_books = train_interactions["Book ID Number"].unique()

# Create mappings
train_user_encoder = {
    user: idx
    for idx, user in enumerate(train_users)
}

train_book_encoder = {
    book: idx
    for idx, book in enumerate(train_books)
}

# Convert IDs to matrix indices
train_user_indices = (
    train_interactions["Membership Number"]
    .map(train_user_encoder)
)

train_book_indices = (
    train_interactions["Book ID Number"]
    .map(train_book_encoder)
)

# Create sparse training matrix
train_matrix = csr_matrix(
    (
        [1] * len(train_interactions),
        (train_user_indices, train_book_indices)
    ),
    shape=(
        len(train_user_encoder),
        len(train_book_encoder)
    )
)

print("===== TRAINING USER-ITEM MATRIX =====")

print("Matrix shape:", train_matrix.shape)
print("Non-zero interactions:", train_matrix.nnz)
print("Number of training users:", len(train_user_encoder))
print("Number of training books:", len(train_book_encoder))
```

```python
from sklearn.metrics.pairwise import cosine_similarity

# Calculate similarity between books
book_similarity = cosine_similarity(
    train_matrix.T,
    dense_output=False
)

print("===== BOOK SIMILARITY MATRIX =====")

print("Shape:", book_similarity.shape)
print("Number of non-zero similarities:", book_similarity.nnz)
```

```python
import numpy as np

def recommend_books(user_id, top_n=10):
    """
    Recommend books for a user using item-based collaborative filtering.
    """

    # Check whether the user exists in training data
    if user_id not in train_user_encoder:
        return None

    # Get the user's encoded index
    user_idx = train_user_encoder[user_id]

    # Get books already borrowed by the user
    user_row = train_matrix[user_idx]

    borrowed_book_indices = user_row.indices

    if len(borrowed_book_indices) == 0:
        return None

    # Accumulate recommendation scores
    scores = {}

    for book_idx in borrowed_book_indices:

        # Get similar books
        similar_books = book_similarity.getrow(book_idx)

        for recommended_idx, similarity_score in zip(
            similar_books.indices,
            similar_books.data
        ):

            # Don't recommend books already borrowed
            if recommended_idx in borrowed_book_indices:
                continue

            # Add similarity score
            scores[recommended_idx] = (
                scores.get(recommended_idx, 0)
                + similarity_score
            )

    # Sort books by recommendation score
    ranked_books = sorted(
        scores.items(),
        key=lambda x: x[1],
        reverse=True
    )

    # Convert encoded book IDs back to original IDs
    reverse_book_encoder = {
        idx: book
        for book, idx in train_book_encoder.items()
    }

    recommendations = []

    for book_idx, score in ranked_books[:top_n]:

        recommendations.append({
            "Book ID Number": reverse_book_encoder[book_idx],
            "Score": round(float(score), 4)
        })

    return recommendations
```

```python
# ==========================================
# STEP 25: Evaluate Item-Based CF
# ==========================================

import numpy as np
import pandas as pd

def evaluate_recommender(test_interactions, k=10):
    precisions = []
    recalls = []

    # Only users that existed during training
    known_users = set(train_interactions["Membership Number"].unique())

    # Only evaluate known users
    test_known = test_interactions[
        test_interactions["Membership Number"].isin(known_users)
    ].copy()

    # Only books known to the training model
    known_books = set(train_interactions["Book ID Number"].unique())

    test_known = test_known[
        test_known["Book ID Number"].isin(known_books)
    ]

    for user_id, group in test_known.groupby("Membership Number"):

        # Actual books borrowed by this user in the test period
        actual_books = set(group["Book ID Number"])

        # Get recommendations
        recs = recommend_books(user_id, top_n=k)

        recommended_books = set(
            rec["Book ID Number"] for rec in recs
        )

        if len(recommended_books) == 0:
            continue

        # Number of correctly recommended books
        hits = len(actual_books & recommended_books)

        precision = hits / k
        recall = hits / len(actual_books)

        precisions.append(precision)
        recalls.append(recall)

    return {
        "Users Evaluated": len(precisions),
        "Precision@{}".format(k): np.mean(precisions),
        "Recall@{}".format(k): np.mean(recalls)
    }


# Evaluate at K = 5
results_5 = evaluate_recommender(
    test_interactions,
    k=5
)

# Evaluate at K = 10
results_10 = evaluate_recommender(
    test_interactions,
    k=10
)

print("===== EVALUATION RESULTS =====")
print()

print("Precision@5 :", round(results_5["Precision@5"], 4))
print("Recall@5    :", round(results_5["Recall@5"], 4))

print()

print("Precision@10:", round(results_10["Precision@10"], 4))
print("Recall@10   :", round(results_10["Recall@10"], 4))

print()
print("Users evaluated:", results_10["Users Evaluated"])
```

```python
# ==========================================
# Evaluate Popularity Baseline
# ==========================================

def evaluate_popularity(test_interactions, k=10):

    precisions = []
    recalls = []

    known_users = set(
        train_interactions["Membership Number"].unique()
    )

    known_books = set(
        train_interactions["Book ID Number"].unique()
    )

    test_known = test_interactions[
        test_interactions["Membership Number"].isin(known_users)
    ].copy()

    test_known = test_known[
        test_known["Book ID Number"].isin(known_books)
    ]

    for user_id, group in test_known.groupby("Membership Number"):

        actual_books = set(group["Book ID Number"])

        recs = popularity_recommend_books(
            user_id,
            top_n=k
        )

        recommended_books = set(
            rec["Book ID Number"]
            for rec in recs
        )

        if not recommended_books:
            continue

        hits = len(
            actual_books & recommended_books
        )

        precision = hits / k
        recall = hits / len(actual_books)

        precisions.append(precision)
        recalls.append(recall)

    return {
        "Users Evaluated": len(precisions),
        f"Precision@{k}": np.mean(precisions),
        f"Recall@{k}": np.mean(recalls)
    }


pop_5 = evaluate_popularity(
    test_interactions,
    k=5
)

pop_10 = evaluate_popularity(
    test_interactions,
    k=10
)

print("\n===== POPULARITY BASELINE =====")

print(
    "Precision@5 :",
    round(pop_5["Precision@5"], 4)
)

print(
    "Recall@5    :",
    round(pop_5["Recall@5"], 4)
)

print(
    "Precision@10:",
    round(pop_10["Precision@10"], 4)
)

print(
    "Recall@10   :",
    round(pop_10["Recall@10"], 4)
)

print(
    "Users evaluated:",
    pop_10["Users Evaluated"]
)
```

```python
# ==========================================
# STEP 27: MATRIX FACTORIZATION WITH SVD
# ==========================================

from sklearn.decomposition import TruncatedSVD
import numpy as np
import pandas as pd

# Use the training interaction matrix created earlier
# train_matrix shape:
# users x books

print("Training matrix shape:", train_matrix.shape)
print("Non-zero interactions:", train_matrix.nnz)

# Number of latent factors
n_components = 50

svd = TruncatedSVD(
    n_components=n_components,
    random_state=42
)

# User x latent-factor matrix
user_factors = svd.fit_transform(train_matrix)

# Book x latent-factor matrix
book_factors = svd.components_.T

print("\n===== SVD MODEL =====")
print("User factors shape:", user_factors.shape)
print("Book factors shape:", book_factors.shape)

print(
    "Explained variance ratio:",
    round(svd.explained_variance_ratio_.sum(), 4)
)
```

```python
# ==========================================
# STEP 28: SVD RECOMMENDATION FUNCTION
# ==========================================

# Reverse mappings
user_to_index = {
    user_id: idx
    for idx, user_id in enumerate(train_interactions["Membership Number"].unique())
}

book_to_index = {
    book_id: idx
    for idx, book_id in enumerate(train_interactions["Book ID Number"].unique())
}

index_to_book = {
    idx: book_id
    for book_id, idx in book_to_index.items()
}


def svd_recommend_books(user_id, top_n=10):

    if user_id not in user_to_index:
        return []

    user_idx = user_to_index[user_id]

    # Predicted preference scores for every book
    scores = user_factors[user_idx].dot(book_factors.T)

    # Books already borrowed during training
    borrowed_books = set(
        train_interactions[
            train_interactions["Membership Number"] == user_id
        ]["Book ID Number"]
    )

    recommendations = []

    # Sort books by predicted score
    ranked_indices = np.argsort(scores)[::-1]

    for book_idx in ranked_indices:

        book_id = index_to_book[book_idx]

        # Don't recommend books already borrowed
        if book_id in borrowed_books:
            continue

        recommendations.append({
            "Book ID Number": book_id,
            "Score": float(scores[book_idx])
        })

        if len(recommendations) >= top_n:
            break

    return recommendations
```

```python
# ==========================================
# STEP 30: EVALUATE SVD
# ==========================================

def evaluate_svd(test_interactions, k=10):

    precisions = []
    recalls = []

    # Users known during training
    known_users = set(
        train_interactions["Membership Number"].unique()
    )

    # Books known during training
    known_books = set(
        train_interactions["Book ID Number"].unique()
    )

    # Keep only known users
    test_known = test_interactions[
        test_interactions["Membership Number"].isin(known_users)
    ].copy()

    # Keep only books known to SVD
    test_known = test_known[
        test_known["Book ID Number"].isin(known_books)
    ]

    for user_id, group in test_known.groupby("Membership Number"):

        # Actual books borrowed by the user in 2025
        actual_books = set(
            group["Book ID Number"]
        )

        # SVD recommendations
        recs = svd_recommend_books(
            user_id,
            top_n=k
        )

        recommended_books = set(
            rec["Book ID Number"]
            for rec in recs
        )

        if len(recommended_books) == 0:
            continue

        # Number of correct recommendations
        hits = len(
            actual_books & recommended_books
        )

        precision = hits / k
        recall = hits / len(actual_books)

        precisions.append(precision)
        recalls.append(recall)

    return {
        "Users Evaluated": len(precisions),
        f"Precision@{k}": np.mean(precisions),
        f"Recall@{k}": np.mean(recalls)
    }


# Evaluate @5
svd_5 = evaluate_svd(
    test_interactions,
    k=5
)

# Evaluate @10
svd_10 = evaluate_svd(
    test_interactions,
    k=10
)


print("===== SVD EVALUATION =====")

print(
    "Precision@5 :",
    round(svd_5["Precision@5"], 4)
)

print(
    "Recall@5    :",
    round(svd_5["Recall@5"], 4)
)

print(
    "Precision@10:",
    round(svd_10["Precision@10"], 4)
)

print(
    "Recall@10   :",
    round(svd_10["Recall@10"], 4)
)

print(
    "Users evaluated:",
    svd_10["Users Evaluated"]
)
```

```python
# ==========================================
# STEP 31.1: PREPARE BOOK CATALOG
# ==========================================

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd
import numpy as np

# One row per book
books = (
    df[
        [
            "Book ID Number",
            "Book Title",
            "Author",
            "DDC Number",
            "Publisher Name",
            "Published Year"
        ]
    ]
    .drop_duplicates(subset=["Book ID Number"])
    .copy()
)

# Fill missing values
books["Book Title"] = books["Book Title"].fillna("")
books["Author"] = books["Author"].fillna("")
books["DDC Number"] = books["DDC Number"].fillna("")
books["Publisher Name"] = books["Publisher Name"].fillna("")
books["Published Year"] = books["Published Year"].fillna("")

# Convert everything to text
books["Book ID Number"] = books["Book ID Number"].astype(str)

# Combine useful metadata
books["content"] = (
    books["Book Title"] + " "
    + books["Author"] + " "
    + books["DDC Number"] + " "
    + books["Publisher Name"]
)

print("Number of books:", len(books))
display(books.head())
```

```python
# ==========================================
# STEP 31.2: TF-IDF REPRESENTATION
# ==========================================

tfidf = TfidfVectorizer(
    lowercase=True,
    stop_words="english",
    ngram_range=(1, 2),
    min_df=2
)

tfidf_matrix = tfidf.fit_transform(
    books["content"]
)

print("TF-IDF matrix shape:", tfidf_matrix.shape)
print("Non-zero values:", tfidf_matrix.nnz)
```

```python
# ==========================================
# STEP 31.3: BOOK CONTENT SIMILARITY
# ==========================================

content_similarity = cosine_similarity(
    tfidf_matrix,
    dense_output=False
)

print(
    "Content similarity shape:",
    content_similarity.shape
)

print(
    "Non-zero similarities:",
    content_similarity.nnz
)
```

```python
# ==========================================
# STEP 31.4: CONTENT-BASED RECOMMENDER
# ==========================================

book_id_to_index = {
    book_id: idx
    for idx, book_id in enumerate(books["Book ID Number"])
}

index_to_book_id = {
    idx: book_id
    for book_id, idx in book_id_to_index.items()
}


def content_recommend_books(user_id, top_n=10):

    # Get books borrowed by the user during training
    user_books = train_interactions[
        train_interactions["Membership Number"] == user_id
    ]["Book ID Number"].astype(str).unique()

    # Keep only books present in content model
    user_books = [
        book_id
        for book_id in user_books
        if book_id in book_id_to_index
    ]

    if len(user_books) == 0:
        return []

    # Accumulate similarity scores
    scores = np.zeros(len(books))

    for book_id in user_books:

        book_idx = book_id_to_index[book_id]

        scores += content_similarity[
            book_idx
        ].toarray().ravel()

    # Average across books the user borrowed
    scores = scores / len(user_books)

    # Books already borrowed
    borrowed_books = set(user_books)

    ranked_indices = np.argsort(scores)[::-1]

    recommendations = []

    for book_idx in ranked_indices:

        book_id = index_to_book_id[book_idx]

        # Don't recommend already borrowed books
        if book_id in borrowed_books:
            continue

        # Ignore zero similarity
        if scores[book_idx] <= 0:
            continue

        recommendations.append({
            "Book ID Number": book_id,
            "Score": float(scores[book_idx])
        })

        if len(recommendations) >= top_n:
            break

    return recommendations
```

```python
# ==========================================
# STEP 32: EVALUATE CONTENT-BASED MODEL
# ==========================================

def evaluate_content(test_interactions, k=10):

    precisions = []
    recalls = []

    known_users = set(
        train_interactions["Membership Number"].unique()
    )

    known_books = set(
        train_interactions["Book ID Number"].unique()
    )

    # Only users known during training
    test_known = test_interactions[
        test_interactions["Membership Number"].isin(known_users)
    ].copy()

    # Only books available to the content model
    test_known = test_known[
        test_known["Book ID Number"].isin(known_books)
    ]

    for user_id, group in test_known.groupby("Membership Number"):

        actual_books = set(group["Book ID Number"])

        recs = content_recommend_books(
            user_id,
            top_n=k
        )

        recommended_books = set(
            rec["Book ID Number"]
            for rec in recs
        )

        if len(recommended_books) == 0:
            continue

        hits = len(
            actual_books & recommended_books
        )

        precision = hits / k
        recall = hits / len(actual_books)

        precisions.append(precision)
        recalls.append(recall)

    return {
        "Users Evaluated": len(precisions),
        f"Precision@{k}": np.mean(precisions),
        f"Recall@{k}": np.mean(recalls)
    }


# Evaluate @5
content_5 = evaluate_content(
    test_interactions,
    k=5
)

# Evaluate @10
content_10 = evaluate_content(
    test_interactions,
    k=10
)

print("===== CONTENT-BASED EVALUATION =====")

print(
    "Precision@5 :",
    round(content_5["Precision@5"], 4)
)

print(
    "Recall@5    :",
    round(content_5["Recall@5"], 4)
)

print(
    "Precision@10:",
    round(content_10["Precision@10"], 4)
)

print(
    "Recall@10   :",
    round(content_10["Recall@10"], 4)
)

print(
    "Users evaluated:",
    content_10["Users Evaluated"]
)
```

```python
# ==========================================
# STEP 36: COLD-START RECOMMENDER
# ==========================================

global_popularity = (
    train_interactions
    .groupby("Book ID Number")["Membership Number"]
    .nunique()
    .sort_values(ascending=False)
)


def cold_start_recommend_books(
    user_id,
    top_n=10
):

    # Find user's profile
    profile = user_profiles[
        user_profiles["Membership Number"] == user_id
    ]

    if len(profile) == 0:
        return []

    profile = profile.iloc[0]

    program = profile["Program Code"]
    department = profile["Department Code"]

    recommendations = []

    # --------------------------------------
    # LEVEL 1: PROGRAM
    # --------------------------------------

    if pd.notna(program):

        program_books = program_book_popularity[
            program_book_popularity["Program Code"]
            == program
        ]

        for _, row in program_books.iterrows():

            recommendations.append({
                "Book ID Number": row["Book ID Number"],
                "Score": float(row["Popularity"]),
                "Source": "Program"
            })

            if len(recommendations) >= top_n:
                return recommendations

    # --------------------------------------
    # LEVEL 2: DEPARTMENT
    # --------------------------------------

    if pd.notna(department):

        department_books = department_book_popularity[
            department_book_popularity["Department Code"]
            == department
        ]

        existing_books = {
            r["Book ID Number"]
            for r in recommendations
        }

        for _, row in department_books.iterrows():

            if row["Book ID Number"] in existing_books:
                continue

            recommendations.append({
                "Book ID Number": row["Book ID Number"],
                "Score": float(row["Popularity"]),
                "Source": "Department"
            })

            if len(recommendations) >= top_n:
                return recommendations

    # --------------------------------------
    # LEVEL 3: GLOBAL POPULARITY
    # --------------------------------------

    existing_books = {
        r["Book ID Number"]
        for r in recommendations
    }

    for book_id, popularity in global_popularity.items():

        if book_id in existing_books:
            continue

        recommendations.append({
            "Book ID Number": book_id,
            "Score": float(popularity),
            "Source": "Global"
        })

        if len(recommendations) >= top_n:
            break

    return recommendations
```

```python
# ============================================
# STEP 38: COLD-START EVALUATION
# ============================================

def evaluate_cold_start_model(top_n=5):

    precisions = []
    recalls = []

    evaluated_users = 0
    total_hits = 0
    total_relevant = 0

    for user_id in cold_start_users:

        # --------------------------------------------
        # 1. Actual books borrowed by this user in 2025
        # --------------------------------------------
        actual_books = set(
            test_interactions[
                test_interactions["Membership Number"] == user_id
            ]["Book ID Number"]
        )

        if len(actual_books) == 0:
            continue

        # --------------------------------------------
        # 2. Generate recommendations
        # --------------------------------------------
        recs = cold_start_recommend_books(
            user_id,
            top_n=top_n
        )

        if recs is None or len(recs) == 0:
            continue

        # --------------------------------------------
        # 3. Extract recommended Book IDs
        # --------------------------------------------

        # Your cold_start_recommend_books() returns a LIST
        # of dictionaries, so extract Book ID Number
        # from each dictionary.

        recommended_books = set()

        for rec in recs:

            if isinstance(rec, dict):
                if "Book ID Number" in rec:
                    recommended_books.add(rec["Book ID Number"])

            else:
                # If the function returns tuples/lists,
                # handle the first element as Book ID.
                try:
                    recommended_books.add(rec[0])
                except:
                    pass

        if len(recommended_books) == 0:
            continue

        # --------------------------------------------
        # 4. Calculate hits
        # --------------------------------------------

        hits = len(
            recommended_books.intersection(actual_books)
        )

        precision = hits / len(recommended_books)
        recall = hits / len(actual_books)

        precisions.append(precision)
        recalls.append(recall)

        total_hits += hits
        total_relevant += len(actual_books)

        evaluated_users += 1

    # --------------------------------------------
    # 5. Return results
    # --------------------------------------------

    return {
        "Precision": (
            sum(precisions) / len(precisions)
            if precisions else 0
        ),

        "Recall": (
            sum(recalls) / len(recalls)
            if recalls else 0
        ),

        "Users Evaluated": evaluated_users,

        "Total Hits": total_hits,

        "Total Relevant": total_relevant
    }


# ============================================
# RUN EVALUATION
# ============================================

cold_start_at5 = evaluate_cold_start_model(top_n=5)

cold_start_at10 = evaluate_cold_start_model(top_n=10)


print("===== COLD-START EVALUATION =====")

print("\nPrecision@5:",
      round(cold_start_at5["Precision"], 4))

print("Recall@5:",
      round(cold_start_at5["Recall"], 4))

print("Users Evaluated:",
      cold_start_at5["Users Evaluated"])


print("\nPrecision@10:",
      round(cold_start_at10["Precision"], 4))

print("Recall@10:",
      round(cold_start_at10["Recall"], 4))

print("Users Evaluated:",
      cold_start_at10["Users Evaluated"])
```

```python
# ============================================
# STEP 42: HYBRID RECOMMENDER
# ============================================

def hybrid_recommend_books(user_id, top_n=10):

    # ----------------------------------------
    # Check whether user has training history
    # ----------------------------------------

    user_has_history = user_id in user_to_index

    # ----------------------------------------
    # COLD-START USER
    # ----------------------------------------

    if not user_has_history:

        return cold_start_recommend_books(
            user_id,
            top_n=top_n
        )


    # ========================================
    # KNOWN USER
    # ========================================

    # Get Item-CF recommendations
    cf_recs = recommend_books(
        user_id,
        top_n=top_n * 3
    )

    # Get SVD recommendations
    svd_recs = svd_recommend_books(
        user_id,
        top_n=top_n * 3
    )

    # ----------------------------------------
    # Store combined scores
    # ----------------------------------------

    scores = {}


    # ========================================
    # ITEM-CF SCORE
    # ========================================

    if cf_recs is not None and len(cf_recs) > 0:

        for rec in cf_recs:

            book_id = rec["Book ID Number"]
            score = rec["Score"]

            scores[book_id] = scores.get(book_id, 0) + (
                0.5 * score
            )


    # ========================================
    # SVD SCORE
    # ========================================

    if svd_recs is not None and len(svd_recs) > 0:

        for rec in svd_recs:

            book_id = rec["Book ID Number"]
            score = rec["Score"]

            scores[book_id] = scores.get(book_id, 0) + (
                0.5 * score
            )


    # ----------------------------------------
    # Sort combined recommendations
    # ----------------------------------------

    ranked_books = sorted(
        scores.items(),
        key=lambda x: x[1],
        reverse=True
    )


    # ----------------------------------------
    # Return top N
    # ----------------------------------------

    recommendations = []

    for book_id, score in ranked_books[:top_n]:

        recommendations.append({
            "Book ID Number": book_id,
            "Score": score,
            "Source": "Hybrid"
        })


    return recommendations
```

```python
# ============================================
# STEP 46: HYBRID EVALUATION
# ============================================

def evaluate_hybrid_model(
    evaluation_users,
    top_n=10
):

    precisions = []
    recalls = []

    total_hits = 0
    total_relevant = 0
    evaluated_users = 0

    for user_id in evaluation_users:

        # ------------------------------------
        # Actual books borrowed in 2025
        # ------------------------------------

        actual_books = set(
            test_interactions[
                test_interactions["Membership Number"] == user_id
            ]["Book ID Number"]
        )

        if len(actual_books) == 0:
            continue

        # ------------------------------------
        # Generate hybrid recommendations
        # ------------------------------------

        recs = hybrid_recommend_books(
            user_id,
            top_n=top_n
        )

        if recs is None or len(recs) == 0:
            continue

        # ------------------------------------
        # Extract recommended Book IDs
        # ------------------------------------

        recommended_books = set()

        for rec in recs:

            if isinstance(rec, dict):

                if "Book ID Number" in rec:
                    recommended_books.add(
                        rec["Book ID Number"]
                    )

            else:

                try:
                    recommended_books.add(rec[0])
                except:
                    pass

        if len(recommended_books) == 0:
            continue

        # ------------------------------------
        # Calculate hits
        # ------------------------------------

        hits = len(
            recommended_books.intersection(
                actual_books
            )
        )

        precision = (
            hits / len(recommended_books)
        )

        recall = (
            hits / len(actual_books)
        )

        precisions.append(precision)
        recalls.append(recall)

        total_hits += hits
        total_relevant += len(actual_books)

        evaluated_users += 1

    # ----------------------------------------
    # Return evaluation results
    # ----------------------------------------

    return {
        "Precision": (
            sum(precisions) / len(precisions)
            if precisions else 0
        ),

        "Recall": (
            sum(recalls) / len(recalls)
            if recalls else 0
        ),

        "Users Evaluated": evaluated_users,

        "Total Hits": total_hits,

        "Total Relevant": total_relevant
    }
```

```python
# ============================================
# STEP 48: NORMALIZED HYBRID RECOMMENDER
# ============================================

def normalize_scores(recommendations):

    if recommendations is None or len(recommendations) == 0:
        return {}

    scores = {}

    for rec in recommendations:

        book_id = rec["Book ID Number"]
        score = float(rec["Score"])

        scores[book_id] = score

    if len(scores) == 0:
        return {}

    min_score = min(scores.values())
    max_score = max(scores.values())

    # Avoid division by zero
    if max_score == min_score:
        return {
            book_id: 1.0
            for book_id in scores
        }

    normalized = {
        book_id:
        (score - min_score) / (max_score - min_score)
        for book_id, score in scores.items()
    }

    return normalized


def normalized_hybrid_recommend_books(
    user_id,
    top_n=10,
    cf_weight=0.5,
    svd_weight=0.5
):

    # ========================================
    # COLD-START USER
    # ========================================

    if user_id not in user_to_index:

        return cold_start_recommend_books(
            user_id,
            top_n=top_n
        )

    # ========================================
    # KNOWN USER
    # ========================================

    cf_recs = recommend_books(
        user_id,
        top_n=top_n * 3
    )

    svd_recs = svd_recommend_books(
        user_id,
        top_n=top_n * 3
    )

    # ========================================
    # NORMALIZE SCORES
    # ========================================

    cf_scores = normalize_scores(cf_recs)
    svd_scores = normalize_scores(svd_recs)

    # ========================================
    # COMBINE
    # ========================================

    combined_scores = {}

    all_books = set(
        cf_scores.keys()
    ).union(
        svd_scores.keys()
    )

    for book_id in all_books:

        cf_score = cf_scores.get(
            book_id,
            0
        )

        svd_score = svd_scores.get(
            book_id,
            0
        )

        combined_score = (
            cf_weight * cf_score
            +
            svd_weight * svd_score
        )

        combined_scores[book_id] = combined_score

    # ========================================
    # RANK
    # ========================================

    ranked_books = sorted(
        combined_scores.items(),
        key=lambda x: x[1],
        reverse=True
    )

    # ========================================
    # TOP-N
    # ========================================

    recommendations = []

    for book_id, score in ranked_books[:top_n]:

        recommendations.append({
            "Book ID Number": book_id,
            "Score": score,
            "Source": "Normalized Hybrid"
        })

    return recommendations
```

```python
# ============================================
# STEP 50: EVALUATE NORMALIZED HYBRID
# ============================================

def evaluate_normalized_hybrid(
    evaluation_users,
    top_n=10
):

    precisions = []
    recalls = []

    evaluated_users = 0

    for user_id in evaluation_users:

        actual_books = set(
            test_interactions[
                test_interactions["Membership Number"] == user_id
            ]["Book ID Number"]
        )

        if len(actual_books) == 0:
            continue

        recs = normalized_hybrid_recommend_books(
            user_id,
            top_n=top_n
        )

        if recs is None or len(recs) == 0:
            continue

        recommended_books = set()

        for rec in recs:

            if isinstance(rec, dict):

                if "Book ID Number" in rec:
                    recommended_books.add(
                        rec["Book ID Number"]
                    )

        if len(recommended_books) == 0:
            continue

        hits = len(
            recommended_books.intersection(
                actual_books
            )
        )

        precision = (
            hits / len(recommended_books)
        )

        recall = (
            hits / len(actual_books)
        )

        precisions.append(precision)
        recalls.append(recall)

        evaluated_users += 1

    return {
        "Precision": (
            sum(precisions) / len(precisions)
            if precisions else 0
        ),

        "Recall": (
            sum(recalls) / len(recalls)
            if recalls else 0
        ),

        "Users Evaluated": evaluated_users
    }


# ============================================
# RUN NORMALIZED HYBRID EVALUATION
# ============================================

normalized_hybrid_at5 = evaluate_normalized_hybrid(
    known_test_users,
    top_n=5
)

normalized_hybrid_at10 = evaluate_normalized_hybrid(
    known_test_users,
    top_n=10
)

print("===== NORMALIZED HYBRID EVALUATION =====")

print(
    "\nPrecision@5:",
    round(
        normalized_hybrid_at5["Precision"],
        4
    )
)

print(
    "Recall@5:",
    round(
        normalized_hybrid_at5["Recall"],
        4
    )
)

print(
    "Users Evaluated:",
    normalized_hybrid_at5["Users Evaluated"]
)

print(
    "\nPrecision@10:",
    round(
        normalized_hybrid_at10["Precision"],
        4
    )
)

print(
    "Recall@10:",
    round(
        normalized_hybrid_at10["Recall"],
        4
    )
)

print(
    "Users Evaluated:",
    normalized_hybrid_at10["Users Evaluated"]
)
```

```python
# ============================================
# STEP 51: COMMON-POPULATION EVALUATION
# ============================================

def evaluate_model_common(
    recommend_function,
    evaluation_users,
    top_n=10
):

    precisions = []
    recalls = []

    evaluated_users = 0

    for user_id in evaluation_users:

        # Actual held-out books
        actual_books = set(
            test_interactions[
                test_interactions["Membership Number"] == user_id
            ]["Book ID Number"]
        )

        if len(actual_books) == 0:
            continue

        # Generate recommendations
        recs = recommend_function(
            user_id,
            top_n=top_n
        )

        if recs is None or len(recs) == 0:
            continue

        recommended_books = set()

        for rec in recs:

            if isinstance(rec, dict):
                if "Book ID Number" in rec:
                    recommended_books.add(
                        rec["Book ID Number"]
                    )

        if len(recommended_books) == 0:
            continue

        # Calculate hits
        hits = len(
            recommended_books.intersection(
                actual_books
            )
        )

        precision = (
            hits / len(recommended_books)
        )

        recall = (
            hits / len(actual_books)
        )

        precisions.append(precision)
        recalls.append(recall)

        evaluated_users += 1

    return {
        "Precision": (
            sum(precisions) / len(precisions)
            if precisions else 0
        ),
        "Recall": (
            sum(recalls) / len(recalls)
            if recalls else 0
        ),
        "Users Evaluated": evaluated_users
    }
```

```python
# ============================================================
# STEP 60.5: TOKENIZE TRAIN / VALIDATION / TEST DATA
# ============================================================

def preprocess_function(examples):
    # Add T5 task prefix
    inputs = ["summarize: " + article for article in examples["article"]]

    # Tokenize articles
    model_inputs = tokenizer(
        inputs,
        max_length=512,
        truncation=True
    )

    # Tokenize summaries as target labels
    labels = tokenizer(
        text_target=examples["highlights"],
        max_length=128,
        truncation=True
    )

    model_inputs["labels"] = labels["input_ids"]

    return model_inputs


print("Tokenizing training data...")
tokenized_train = train_small.map(
    preprocess_function,
    batched=True,
    remove_columns=train_small.column_names
)

print("Tokenizing validation data...")
tokenized_validation = validation_small.map(
    preprocess_function,
    batched=True,
    remove_columns=validation_small.column_names
)

print("Tokenizing test data...")
tokenized_test = test_small.map(
    preprocess_function,
    batched=True,
    remove_columns=test_small.column_names
)

print("\nTokenization completed successfully.")

print("\nDataset sizes:")
print("Train:", len(tokenized_train))
print("Validation:", len(tokenized_validation))
print("Test:", len(tokenized_test))

print("\nTokenized columns:")
print(tokenized_train.column_names)

print("\nFirst training example:")
print("Input token count:", len(tokenized_train[0]["input_ids"]))
print("Label token count:", len(tokenized_train[0]["labels"]))
```

```python
# ============================================================
# STEP 60.13: TRAIN T5-SMALL FOR 1 EPOCH
# ============================================================

import time
import torch

# ------------------------------------------------------------
# Training setup
# ------------------------------------------------------------

torch.manual_seed(RANDOM_SEED)

model.train()

total_batches = len(train_dataloader)
total_steps = total_batches // GRADIENT_ACCUMULATION_STEPS

running_loss = 0.0
epoch_loss = 0.0
optimizer.zero_grad()

start_time = time.time()

print("Starting T5-small training...")
print("=" * 60)
print("Training examples:", len(tokenized_train))
print("Epochs:", NUM_EPOCHS)
print("Batch size:", TRAIN_BATCH_SIZE)
print("Gradient accumulation:", GRADIENT_ACCUMULATION_STEPS)
print("Optimization steps:", total_steps)
print("Device:", device)
print("=" * 60)

# ------------------------------------------------------------
# Training loop
# ------------------------------------------------------------

for batch_idx, batch in enumerate(train_dataloader):

    # Move batch to MPS
    batch = {
        key: value.to(device)
        for key, value in batch.items()
    }

    # Forward pass
    outputs = model(**batch)

    loss = outputs.loss

    # Scale loss for gradient accumulation
    scaled_loss = loss / GRADIENT_ACCUMULATION_STEPS

    # Backward pass
    scaled_loss.backward()

    running_loss += loss.item()
    epoch_loss += loss.item()

    # Optimizer update
    if (
        (batch_idx + 1) % GRADIENT_ACCUMULATION_STEPS == 0
        or (batch_idx + 1) == total_batches
    ):

        # Gradient clipping
        torch.nn.utils.clip_grad_norm_(
            model.parameters(),
            max_norm=1.0
        )

        optimizer.step()
        optimizer.zero_grad()

        current_step = (
            (batch_idx + 1) // GRADIENT_ACCUMULATION_STEPS
        )

        # Print progress every 100 optimization steps
        if current_step % 100 == 0 or current_step == total_steps:

            avg_loss = running_loss / (
                100 * GRADIENT_ACCUMULATION_STEPS
            )

            elapsed = time.time() - start_time

            print(
                f"Step {current_step:4d}/{total_steps} | "
                f"Avg loss: {avg_loss:.4f} | "
                f"Time: {elapsed/60:.1f} min"
            )

            running_loss = 0.0

# ------------------------------------------------------------
# Training summary
# ------------------------------------------------------------

average_training_loss = epoch_loss / total_batches

elapsed_total = time.time() - start_time

print("\n" + "=" * 60)
print("TRAINING COMPLETED")
print("=" * 60)

print("Average training loss:", round(average_training_loss, 4))
print("Total training time:", round(elapsed_total / 60, 2), "minutes")
print("Optimization steps:", total_steps)
```

```python
# ============================================================
# STEP 60.14: CALCULATE VALIDATION LOSS
# ============================================================

import torch
import time

model.eval()

validation_loss = 0.0
validation_batches = 0

start_time = time.time()

print("Evaluating validation dataset...")
print("=" * 60)

with torch.no_grad():

    for batch in validation_dataloader:

        # Move batch to MPS
        batch = {
            key: value.to(device)
            for key, value in batch.items()
        }

        # Forward pass
        outputs = model(**batch)

        validation_loss += outputs.loss.item()
        validation_batches += 1

validation_loss = validation_loss / validation_batches

elapsed = time.time() - start_time

print("\nValidation completed.")
print("=" * 60)
print("Validation batches:", validation_batches)
print("Average validation loss:", round(validation_loss, 4))
print("Evaluation time:", round(elapsed / 60, 2), "minutes")
```
